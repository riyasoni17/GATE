```python
# def remdup(li):
#     new=li
#     count=0
#     for i in range(0,len(li)):
#         for j in range(0,len(li)):
#             if li[i]==li[j]:
#                 new.pop(i-count)
#                 count=count+1
#     return new


def remdup(l):
    seen = set()
    new_list = []
    for item in reversed(l):
        if item not in seen:
            seen.add(item)
            new_list.append(item)
    return new_list[::-1]

```

```python
# Test cases
print(remdup([3, 1, 3, 5]))  # Output: [1, 3, 5]
print(remdup([7, 3, -1, -5]))  # Output: [7, 3, -1, -5]
print(remdup([3, 5, 7, 5, 3, 7, 10]))  # Output: [5, 3, 7, 10]
```

```python
def splitsum(l):
    pos=0
    neg=0
    for i in l:
        if i>=0:
            pos=pos+(i**2)
        else:
            neg=neg+(i**3)
    return [pos,neg]
```

```python
splitsum([1,3,-5])
```

```python
splitsum([2,4,6])
```

```python
splitsum([-19,-7,-6,0])
```

```python
splitsum([-1,2,3,-7])
```

```python
def matrixflip(m,d):
    if d=='h':
        return [i[::-1] for i in m]
    elif d=='v':
        return m[::-1]
    else: 
        return m
```

```python
myl = [[1,2],[3,4]]
myl
```

```python
matrixflip(myl,'h')
# [[2, 1], [4, 3]]
```

```python
myl
# [[1, 2], [3, 4]]  
```

```python
matrixflip(myl,'v')
# [[3, 4], [1, 2]]  
```
