```python
def mystery(l):
  if (l == []):
    return(l)
  else:
    mid = len(l)//2
    if (len(l)%2 == 0):
      return l[mid-1:mid+1] + mystery(l[:mid-1]+l[mid+1:])
    else:
      return l[mid:mid+1] + mystery(l[:mid]+l[mid+1:])
```

```python
mystery([22,14,19,65,82,55])
```

```python
mystery([22,14,19,65,82])
```

```python
def rainaverage(l):
    d={}
    for i in d:
        if i[0] in d:
            d[i[0]]+=i[1]
        else:
            d[i[0]]+=i[1]
    for i in d:
        d[i]=d[i]/len(l)
    return d
```

```python
rainaverage([(1,2),(1,3),(2,3),(1,1),(3,8)])
```
